```python
# INTEGER TO FLOAT CONVERSION 
a=20
print(type (a)) # TYPE() ka use data type check krne ke liye
print(a) 
b = float (a)
print(b)
print(type (b))
```

```python
# FLOAT TO INTEGER CONVERSION 
a = 10.5 
print(type (a))
print(a)
b = int (a)
print(b)
print(type (b))
```

```python
# Boolean to integer and float 

# true conversion 
a= True # true ka data typr bool h --
print(type (a ))
print(a)

b = float (a)
print(b)
print(type (b))

b = int (a)
print(b)
print(type (b))

# FALSE conversion

a = False # false ka data type bool h
print( type ( a))
print(a)

b = float (a)
print(a)
print(type (b))

b = int (a)
print(b)
print(type (b))

```

```python
# compex number 
x = 5 
y = 9 

z = complex (x,y)
print(type (z))
print(z)

print(z.real)
print(z.imag)
```

```python
# direct method 
a = 3 + 9j 
b = 1 + 5j

print(type (a))
print(type (b))

print(a+b)
print(a-b)
print(a.real)
print(a.imag)
```

```python
# None type
a = None 
print(type (a ))
print(a)

# String to float conversion

a = "20.5"
print(type(a))
print(a) 

b = float (a)
print(b)
print(type (b))

# string to integer conversion 

a = "20"
print(type(a))
print(a)

b = int (a)
print(type (b))
print(b)
```

```python
# STRING DATA TYPES     # INDEXING IN STRING 
 
str1 = "welcome"
print(str1[3]) # forward indexing 
print(str1[-4]) # backward indexing
print(str1 [-1])
print(str1[-7])
```
