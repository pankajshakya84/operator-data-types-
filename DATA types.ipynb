```python
# INTEGER TO FLOAT CONVERSION 
a=20
print(type (a)) # TYPE() ka use data type check krne ke liye
print(a) 
b = float (a)
print(b)
print(type (b))
```

```python
# FLOAT TO INTEGER CONVERSION 
a = 10.5 
print(type (a))
print(a)
b = int (a)
print(b)
print(type (b))
```

```python
# Boolean to integer and float 

# true conversion 
a= True # true ka data typr bool h --
print(type (a ))
print(a)

b = float (a)
print(b)
print(type (b))

b = int (a)
print(b)
print(type (b))

# FALSE conversion

a = False # false ka data type bool h
print( type ( a))
print(a)

b = float (a)
print(a)
print(type (b))

b = int (a)
print(b)
print(type (b))

```

```python
# compex number 
x = 5 
y = 9 

z = complex (x,y)
print(type (z))
print(z)

print(z.real)
print(z.imag)
```

```python
# direct method 
a = 3 + 9j 
b = 1 + 5j

print(type (a))
print(type (b))

print(a+b)
print(a-b)
print(a.real)
print(a.imag)
```

```python
# None type
a = None 
print(type (a ))
print(a)

# String to float conversion

a = "20.5"
print(type(a))
print(a) 

b = float (a)
print(b)
print(type (b))

# string to integer conversion 

a = "20"
print(type(a))
print(a)

b = int (a)
print(type (b))
print(b)
```

```python
# .count(string,begin,en) :- begin se end tak diye gaye character ya string ko count krta h 
str1 = "welcome to python programmeng"
print(str1.count("o"))
print(str1.count("me"))
print(str1.count("o",6))
print(str1.count("o",6,10))
```

```python
# .find(string, begin , end) :- yh string find krta h [ yh string na milne pr error nhi dega (-1) ayega ]
str1 = "welcome to python programming"
print(str1.find("come"))
print(str1.find("to"))
print(str1.find("wel"))
print(str1.find("python"))
print(str1.find("come",20))
print(str1.find("wel",5))
```

```python
# .index(string, begin , end) :- yh string find krta h [ yh string na milne pr error nhi dega (error) ayega ]
str1 = "welcome to python programming"
print(str1.index("come"))
print(str1.index("to"))
print(str1.index("wel"))
print(str1.index("python"))
print(str1.index("come",20))
print(str1.index("wel",5))
```
