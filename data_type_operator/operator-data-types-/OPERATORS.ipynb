```python
# ARITHMETICAL OPERATORS

a = 10
b = 20

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b) # MODULAS remainder batata h 
print(a ** b) # exponent powere yh power lgata h 
print(a // b) # floor division decimal value hata kr lower integer deta h 
```

```python
# comparision  OPERATORS #  YH VALUE ASSIGN KRNE KE LIYE USE HOTA H 
a = 10
b = 20

print(a == b) # DONO Value check krta h baraber h ya na 
print(a != b) # dono value brabar nhi h kya yh check krta h 
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
# COMPARISION OPERATORS # YH VALUE COMPARISION KRNE KE LIYE USE HOTA H 
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
# LOGICAL OPERATORS 
a = 10
b = 20

print(a < b and b > a) # AND DONO CONDITION TRUE HOGI TBHI TRUE DEGA 
print(a < b and b < a)
print(a < b or b < a) # KOI EK CONDITION  TRUE HOGI TO TRUE AYEGA 
print(a > b or b < a)
```

```python
a = 10
b = 20
c = 30

print(a < b and b > a and c < b)
print(a < b and b > a and c > a)
```

```python
a = 10
b = 20

print(a == b)
print(not(a == b)) # NOT KA KAM HOTA H REVERSE KRNA 

print(a != b)
print(not(a != b))
```

```python
# MEMBERSHIP OPERATOR IN OR NOT IN CHECK KRTA H KI VALUE LIST YA STRING ME PRESENT HAI YA NA 
a = [10, 20, 30, 40, 50, 70]

print(30 in a) # IN KA USE VALUE LIST ME H 
print(700 in a)
print(600 not in a) # NOT IN KA USE VALUE LIST ME NHI H 

str1 = "Welcome"

print("come" in str1)
print("wel" in str1)
print("Wel" in str1)
print("ll" in str1)
print("Me" not in str1)
```

```python
# IDENTIFY OPERATORS 
a = 10
b = 20

print(a is b)     # IS check krta h dono value  same h kya 
print(a is not b) # IS NOT  check krta h dono value same nhi h kya 

str1 = "Hello"
str2 = "hello"

print(str1 is str2)
print(str1 is not str2)

print(a == b)
print(a != b)

print(str1 == str2)
```

```python
# WALRUS OPERATORS 
a = 20.5 # value assign hue 
print(a)
print(a:= 20.5) # ek hi line me value assign bhi hue or print bhi hue 
```

```python
# BITWISE OPERATORS 
print(False and False) 
print(True and False)
print(False and True)
print(True and True)
```

```python
print(bool(0 and 0)) # AND me dono value true hogi tbhi true hogi 
print(bool(1 and 1))
print(bool(0 and 1))
print(bool(1 and 0))
print(bool(19 and 10))
```

```python
print(bool(0 or 0)) # OR me agr ek bhi value true hogi to true ayega 
print(bool(1 or 0))
print(bool(0 or 1))
print(bool(1 or 1))
```

```python
print(bool(0 ^ 0)) # XOR me  sirf ek value hi alg hogi  tbhi to  true ayega 
print(bool(1 ^ 0))
print(bool(0 ^ 1))
print(bool(1 ^ 1))
```
